## Helpers

```python
def expand_box(x1, y1, x2, y2, w, h, border):
    """Expand [x1,y1,x2,y2] by `border` pixels, staying inside image size (w,h)."""
    x1b = max(0, int(x1) - border)
    y1b = max(0, int(y1) - border)
    x2b = min(w - 1, int(x2) + border)
    y2b = min(h - 1, int(y2) + border)
    return x1b, y1b, x2b, y2b


def process_video(video_path, species):
    """Run megafishdetector on one video and save border crops."""
    cap = cv2.VideoCapture(str(video_path))
    if not cap.isOpened():
        print("  Could not open:", video_path)
        return

    species_dir = CROP_ROOT / species
    species_dir.mkdir(parents=True, exist_ok=True)

    frame_idx = 0
    saved_count = 0
    video_stem = video_path.stem

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        h, w = frame.shape[:2]

        # YOLO inference (expects np.array BGR is fine)
        results = model(frame)

        # results.xyxy[0] : (N, 6)  -> x1,y1,x2,y2,conf,cls
        dets = results.xyxy[0].cpu().numpy()

        for det_i, det in enumerate(dets):
            x1, y1, x2, y2, conf, cls_id = det

            # Expand box with border, clip to image
            bx1, by1, bx2, by2 = expand_box(x1, y1, x2, y2, w, h, BORDER_PX)

            crop = frame[by1:by2, bx1:bx2]
            if crop.size == 0:
                continue

            # Save crop; you can include class id if you want
            out_name = "{}_f{:05d}_d{:02d}.jpg".format(
                video_stem, frame_idx, det_i
            )
            out_path = species_dir / out_name
            cv2.imwrite(str(out_path), crop)
            saved_count += 1

        frame_idx += 1

    cap.release()
    print("  Saved {} crops from {}".format(saved_count, video_path.name))
```

```python
def main():
    if not VIDEO_ROOT.exists():
        print("VIDEO_ROOT does not exist:", VIDEO_ROOT)
        return

    species_dirs = [d for d in VIDEO_ROOT.iterdir() if d.is_dir()]
    print("Found {} species folders under {}".format(len(species_dirs), VIDEO_ROOT))

    for species_dir in sorted(species_dirs):
        species = species_dir.name
        mp4_files = sorted(species_dir.glob("*.mp4"))

        print("\nSpecies: {}  ({} clips)".format(species, len(mp4_files)))
        for vid_path in mp4_files:
            print(" Processing:", vid_path.relative_to(VIDEO_ROOT))
            try:
                process_video(vid_path, species)
            except Exception as e:
                print("  ERROR on {}: {}".format(vid_path, e))


if __name__ == "__main__":
    main()
```

## UMAP

```python
from pathlib import Path

import numpy as np
from PIL import Image
import cv2
import torch
from transformers import AutoImageProcessor, AutoModel
import umap
import matplotlib.pyplot as plt
import pandas as pd
```

```python
CROP_ROOT = Path("/Users/wendycao/fish/processed/fish_crops_border")
MODEL_NAME = "facebook/dinov2-base"   # you can try dinov2-large later

```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)

print("Loading DINOv2...")
image_processor = AutoImageProcessor.from_pretrained(MODEL_NAME)
dinov2_model = AutoModel.from_pretrained(MODEL_NAME).to(device)
dinov2_model.eval()
print("DINOv2 loaded.")
```

```python
def dinov2_embed_image(img):
    """Return a 1D DINOv2 embedding from a PIL image."""
    inputs = image_processor(images=img, return_tensors="pt").to(device)
    with torch.no_grad():
        outputs = dinov2_model(**inputs)
        cls_emb = outputs.last_hidden_state[:, 0, :]  # CLS token
    return cls_emb.squeeze(0).cpu().numpy()


def dinov2_embed_crop(image_path):
    """Load crop image and return DINOv2 embedding."""
    img = Image.open(image_path).convert("RGB")
    return dinov2_embed_image(img)
```

```python
if not CROP_ROOT.exists():
    raise SystemExit("CROP_ROOT does not exist: {}".format(CROP_ROOT))

embeddings = []
species_labels = []
crop_paths = []

species_dirs = [d for d in CROP_ROOT.iterdir() if d.is_dir()]
print("Found {} species folders under {}".format(len(species_dirs), CROP_ROOT))

for species_dir in sorted(species_dirs):
    species = species_dir.name
    img_files = sorted(species_dir.glob("*.jpg"))

    print("\nSpecies: {}  ({} crops)".format(species, len(img_files)))
    for i, img_path in enumerate(img_files):
        rel = img_path.relative_to(CROP_ROOT)
        print("  [{}/{}] {}".format(i + 1, len(img_files), rel))

        try:
            emb = dinov2_embed_crop(str(img_path))
        except Exception as e:
            print("    -> FAILED to embed (skipping):", e)
            continue

        embeddings.append(emb)
        species_labels.append(species)
        crop_paths.append(str(rel))

embeddings = np.stack(embeddings)
print("\nCollected embeddings:", embeddings.shape)
```

```python
print("Running UMAP on crop embeddings...")
reducer = umap.UMAP(
    n_neighbors=15,
    min_dist=0.1,
    metric="cosine",
    random_state=42,
)
emb_2d = reducer.fit_transform(embeddings)
print("UMAP done. emb_2d shape:", emb_2d.shape)
```

```python
if emb_2d.shape[0] == 0:
    print("No points to plot.")
else:
    unique_species = sorted(set(species_labels))
    species_to_idx = {s: i for i, s in enumerate(unique_species)}
    color_indices = np.array([species_to_idx[s] for s in species_labels])

    plt.figure(figsize=(8, 6))
    scatter = plt.scatter(
        emb_2d[:, 0],
        emb_2d[:, 1],
        c=color_indices,
        s=10,
        alpha=0.8,
        cmap="tab10",
    )
    plt.title("UMAP of DINOv2 embeddings (fish crops with border)")
    plt.tight_layout()

    out_png = CROP_ROOT.parent / "dinov2_crops_umap.png"
    plt.savefig(out_png, dpi=300)
    print("Saved UMAP plot to:", out_png)

    plt.show()


out_dir = CROP_ROOT.parent / "dinov2_crops_embeddings"
out_dir.mkdir(exist_ok=True)

np.save(out_dir / "embeddings.npy", embeddings)

meta_df = pd.DataFrame(
    {
        "crop_path": crop_paths,
        "species": species_labels,
    }
)
meta_df.to_csv(out_dir / "metadata.csv", index=False)

print("Saved crop embeddings to:", out_dir / "embeddings.npy")
print("Saved crop metadata to:", out_dir / "metadata.csv")
```
